```python
import math
longueur = int(input("Entrez la longueur du rectangle : "))
largeur = int(input("Entrez la largeur du rectangle : "))

perimetre = 2 * (longueur + largeur)
surface = longueur * largeur

d = longueur**2 + largeur**2

diagonale = math.sqrt(d)

print(f"Le perimetre du rectangle est : {perimetre}")
print(f"La surface du rectangle est : {surface}")
print(f"La diagonale du rectangle est : {diagonale: .2f}")
```
